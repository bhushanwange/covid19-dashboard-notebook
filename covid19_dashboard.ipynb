# COVID-19 Summary Dashboard Notebook
This notebook presents a multi-chart visual summary of COVID-19 confirmed cases, regional trends, monthly patterns, and top-affected countries[cite: 7].

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load local COVID-19 dataset
df = pd.read_csv("covid19_data.csv")
df["date"] = pd.to_datetime(df["date"])
df["month"] = df["date"].dt.to_period("M").astype(str)

print("Dataset Shape:", df.shape)
display(df.head())
```

### Visualization 1: Time-Series Trend of Confirmed Cases Over Time
Tracks total confirmed cases across recorded dates with clear labels and legend[cite: 7].

```python
daily_cases = df.groupby("date", as_index=False)[["confirmed", "deaths"]].sum()

plt.figure(figsize=(9, 4.5))
plt.plot(daily_cases["date"], daily_cases["confirmed"], marker="o", linewidth=2, label="Confirmed Cases")
plt.plot(daily_cases["date"], daily_cases["deaths"], marker="s", linewidth=2, label="Deaths")
plt.title("COVID-19 Confirmed Cases & Deaths Over Time", fontsize=13)
plt.xlabel("Date")
plt.ylabel("Number of Cases")
plt.legend(title="Metric")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

### Visualization 2: Heatmap of Confirmed Cases by Region and Month
Compares case intensity across regions across each month[cite: 7].

```python
region_month_pivot = df.pivot_table(
    index="region",
    columns="month",
    values="confirmed",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(8, 4.5))
sns.heatmap(region_month_pivot, annot=True, fmt=",.0f", cmap="YlOrRd", cbar_kws={"label": "Confirmed Cases"})
plt.title("Confirmed Cases Heatmap by Region and Month", fontsize=13)
plt.xlabel("Month")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

### Visualization 3: Top 10 Most Affected Countries
Ranks the top 10 countries by peak/latest cumulative confirmed cases[cite: 7].

```python
top_10_countries = (
    df.groupby("country", as_index=False)["confirmed"]
    .max()
    .sort_values(by="confirmed", ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 4.5))
sns.barplot(data=top_10_countries, x="country", y="confirmed", palette="viridis", hue="country", legend=False)
plt.title("Top 10 Countries by Confirmed COVID-19 Cases", fontsize=13)
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Visualizations 4 & 5: Regional Share (Horizontal Bar Chart) & Case Distribution (Box Plot)
Adds two additional distinct chart types to complete a five-visualization dashboard summary[cite: 7].

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Visualization 4: Horizontal bar chart of confirmed cases by region
region_totals = df.groupby("region", as_index=False)["confirmed"].sum().sort_values(by="confirmed", ascending=True)
axes[0].barh(region_totals["region"], region_totals["confirmed"], color="teal")
axes[0].set_title("Total Confirmed Cases by Region", fontsize=12)
axes[0].set_xlabel("Confirmed Cases")
axes[0].set_ylabel("Region")

# Visualization 5: Box plot of confirmed case spread by region
sns.boxplot(data=df, x="region", y="confirmed", ax=axes[1], palette="Set2", hue="region", legend=False)
axes[1].set_title("Distribution of Confirmed Cases by Region", fontsize=12)
axes[1].set_xlabel("Region")
axes[1].set_ylabel("Confirmed Cases")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```

## Summary of Dashboard Insights
* **Rapid Escalation:** Confirmed cases grew sharply between March and May across tracked regions[cite: 7].
* **Regional Concentration:** Europe and the Americas accounted for high case volumes in the sample dataset, with the USA leading individual country totals[cite: 7].
* **Multi-View Clarity:** Combining line trends, heatmaps, bar rankings, and distribution plots highlights both overall growth and regional variation[cite: 7].
